```python
# Imports, taken from Lab2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

Q1

```python
Grc_df = pd.read_csv('./Grocery_dataset.csv', index_col=0) # Loads data from csv file in 'Grc_df'

print("DataFrame information:") # Prints DataFrame information, showing that there are 5000 rows and 11 columns.
display(Grc_df.shape)

print("\nFirst 5 rows:") # Remaining lines print first 5 and last 5 rows of the dataset.
display(Grc_df.head())

print("\nLast 5 rows:")
display(Grc_df.tail())
```

Q2

```python
null_Grc_df = Grc_df.copy()

print('Null values:') # Prints number of null values by category
display(null_Grc_df.isnull().sum())

drop_Grc_df = Grc_df.copy() # Copies DataFrame and removes null values using dropna() method
drop_Grc_df.dropna(inplace=True)

drop_Grc_df.reset_index(inplace=True) # Reset and set index to default
drop_Grc_df.set_index('Item_Identifier', inplace=True)

print('\nTBA:') # Prints first five rows, current number of rows, and number of null values by cateogry (Now 0)
display(drop_Grc_df.head())
display(drop_Grc_df.shape)
display(drop_Grc_df.isnull().sum())
```

Q3

```python
print('Unique outlet sizes:') # Prints number of unique outlet sizes (null values included)
display(len(Grc_df['Outlet_Size'].unique()))

print('\nMaximum and minimum outlet size:') # Print set of maximums and minimums. Surprisingly, Small is maximum and High is minimum.
display(Grc_df[Grc_df['Outlet_Size'] == Grc_df['Outlet_Size'].max()]['Outlet_Size'])
display(Grc_df[Grc_df['Outlet_Size'] == Grc_df['Outlet_Size'].min()]['Outlet_Size'])
```

```python
print('List of unique Item Fat Content types:') # Showing uniqe in total, and listing them
print('Count: ' + str(len(Grc_df['Item_Fat_Content'].unique())))
display(Grc_df['Item_Fat_Content'].value_counts())

Grc_df['Item_Fat_Content'] = Grc_df['Item_Fat_Content'].replace({ # Replacing names logically and printing the new list
    'LF': 'Low Fat',
    'low fat': 'Low Fat',
    'reg': 'Regular'
})

display(Grc_df['Item_Fat_Content'].value_counts())
```

Q5

```python
Grc_new_df = Grc_df.drop(columns=Grc_df.columns[[0, 6]]) # Creates new DataFrame, drops index values 0 and 6

display(Grc_new_df.head()) # Prints both DataFrames showing that 'item_weight' and 'Outlet_Establishment_Year' are dropped
display(Grc_df.head()) 
```

Q6

```python
display(Grc_df['Outlet_Type'])
```

```python
SubType_1 = Grc_new_df[Grc_new_df['Outlet_Type'] == 'Supermarket Type1'].copy()
SubType_2 = Grc_new_df[Grc_new_df['Outlet_Type'] == 'Supermarket Type2'].copy()

display(SubType_1.head())
display(SubType_2.head())
```

Q7

```python
plt.style.use('ggplot') # Applies ggplot style to the plots

fig, axes = plt.subplots(2, 1, figsize=(8, 10))

sns.boxplot(x=SubType_1['Item_MRP'], ax=axes[0]) # 5-point summary boxplot of Item_MRP for Supermarket Type1
axes[0].set_title('Item_MRP - Supermarket Type1')

sns.boxplot(x=SubType_2['Item_MRP'], ax=axes[1]) # 5-point summary boxplot of Item_MRP for Supermarket Type2
axes[1].set_title('Item_MRP - Supermarket Type2')

plt.tight_layout()
plt.show()

print('Median Item_MRP, Supermarket Type1:', SubType_1['Item_MRP'].median())
print('Median Item_MRP, Supermarket Type2:', SubType_2['Item_MRP'].median())
```

Q8

```python
Grc_Concat_df = pd.concat([SubType_1 , SubType_2 ], ignore_index=True) # Concatinates DataFrames into new DataFrame

Grc_Concat_df.sort_values(by='Item_Outlet_Sales') # Prints concatinated DataFrame, showing Tier 3 having lowest sales.
```
