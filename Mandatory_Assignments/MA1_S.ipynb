```python
# Imports, taken from Lab2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

Q1

```python
# Loads data from csv file in 'Grc_df'
Grc_df = pd.read_csv('MA1/Grocery_dataset.csv') 

# Prints DataFrame information, showing that there are 5000 rows and 11 columns.
print("DataFrame information:") 
display(Grc_df.shape)

# Prints first 10 and last 10 rows of the dataset.
print("\nFirst 10 rows:") 
display(Grc_df.head(10))

print("\nLast 10 rows:")
display(Grc_df.tail(10))
```

Q2

```python
# Prints number of null values by category
print('Null values:')
display(Grc_df.isnull().sum())

# Removes null values using dropna() method
Grc_df.dropna(inplace=True)

# Prints first five rows, current number of rows, and number of null values by cateogry (Now 0)
print('\nDataFrame without null values:')
display(Grc_df.head())
display(Grc_df.shape)
display(Grc_df.isnull().sum())
```

Q3

```python
# Prints number of unique outlet sizes
print('Unique outlet sizes:') 
display(Grc_df['Outlet_Size'].nunique())

# Prints maximum and minimum values.
print('Maximum value:')
display(Grc_df['Outlet_Size'].max())
print('Minimum value:')
display(Grc_df['Outlet_Size'].min())

print('Maximum value is small while Minimum value is High. \nThis is because the column of Outlet Size is categorical \nwhile values are sorted alphabetically.')
```

Q4

```python
# Showing unique in total, and listing them
print('List of unique Item Fat Content types:')
print('Count: ' + str(len(Grc_df['Item_Fat_Content'].unique())))
display(Grc_df['Item_Fat_Content'].value_counts())

# Replacing names logically and printing the new list
Grc_df['Item_Fat_Content'] = Grc_df['Item_Fat_Content'].replace({ 
    'LF': 'Low Fat',
    'low fat': 'Low Fat',
    'reg': 'Regular'
})

display(Grc_df['Item_Fat_Content'].value_counts())
```

Q5

```python
# Creates new DataFrame, drops index values 0 and 6
Grc_new_df = Grc_df.drop(columns=Grc_df.columns[[0, 6]])

# Prints both DataFrames showing that 'Item_Identifier' and 'Outlet_Identifier' are removed
display(Grc_new_df.head()) 
display(Grc_df.head()) 
```

Q6

```python
# Creates two new copies of the DataFrame, each containing only one of the types of supermarkets
SupType_1 = Grc_new_df[Grc_new_df['Outlet_Type'] == 'Supermarket Type1'].copy()
SupType_2 = Grc_new_df[Grc_new_df['Outlet_Type'] == 'Supermarket Type2'].copy()
```

Q7

```python
# Applies ggplot style to the plots
plt.style.use('ggplot')

# 5-point summary boxplot of Item_MRP for Supermarket Type1 and Type2
plt.figure(figsize= (10,15))
plt.subplot(2,1,1)
sns.boxplot(x='Outlet_Type', y=SupType_1['Item_MRP'], color='blue', data=SupType_1)
plt.subplot(2,1,2)
sns.boxplot(x='Outlet_Type', y=SupType_2['Item_MRP'], color='red', data=SupType_2) 

# SupType_1 has the highest median, while none of them have any outliers.
```

Q8

```python
# Concatinates DataFrames into new DataFrame
Grc_Concat_df = pd.concat([SupType_1 , SupType_2 ], ignore_index=True) 

# Prints concatinated DataFrame, showing Tier 3 having lowest sales.
Grc_Concat_df.sort_values(by='Item_Outlet_Sales')
```

Q9

```python
# Creates multiple indexes from size and location type
Grc_Concat_df = Grc_Concat_df.set_index( 
    ['Outlet_Size', 'Outlet_Location_Type']
)

# Prints multiple indexed DataFrame
display(Grc_Concat_df.head()) 
```

Q10

```python
# Divides Item_Weight into 10 equal sized buckets
weight_buckets = pd.cut(Grc_Concat_df['Item_Weight'], bins=10) 

# Groups the buckets and calculates mean, min, max, and count for each bucket
bucket_stats = Grc_Concat_df['Item_Weight'].groupby(weight_buckets).agg( 
    Mean='mean',
    Minimum='min',
    Maximum='max',
    Count='count'
)

# Prints stats per bucket
display(bucket_stats) 
```
