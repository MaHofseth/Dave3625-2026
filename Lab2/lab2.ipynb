```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/DAVE3625/Dave3625-2026/refs/heads/main/Lab2/data/stud.csv"
df = pd.read_csv(url, sep=',')
df.head()
```

```python
df.info()
```

```python
df = df.replace(r'^\s*$', np.nan, regex=True)  # blanks -> NaN
df = df.dropna()                                # remove rows with NaN

df['Age'] = df['Age'].astype('int32')
df['hrsStudy'] = df['hrsStudy'].astype('int32')
```

```python
df = df[df["FinalGrade"] <= 100]
df["FinalGrade"].plot.box()
```

```python
conditions = [
    df["FinalGrade"] > 90,
    df["FinalGrade"] > 80,
    df["FinalGrade"] > 70,
    df["FinalGrade"] > 60,
    df["FinalGrade"] > 50,
]
letters = ["A", "B", "C", "D", "E"]

df["Grade"] = np.select(conditions, letters, default="F")
```

```python
df["Grade"].value_counts().sort_index().plot.bar()
plt.show()
```
